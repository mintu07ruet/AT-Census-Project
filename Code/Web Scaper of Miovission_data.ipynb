```python
url = 'https://datalink.miovision.com/'
exceptlist=[]
from selenium.common.exceptions import NoSuchElementException

# make a GET request
for page in range(11,12):
    urls = 'https://datalink.miovision.com/studies?page=' + str(page)

    browser.get(urls)
    time.sleep(150)
    html_page = browser.page_source
    from bs4 import BeautifulSoup
    soup = BeautifulSoup(html_page)
    studyids=soup("tr", class_="marker_hover leaf collapsed")
    #print(studyids)
    type(studyids)
    study_nums = [item["data-marker-value"] for item in studyids]
    #print(study_nums)
    print(study_nums)
    for i in range(0,len(study_nums)):
        url2=url+'studies/' + study_nums[i] + '/report'
        # soup.find_all("a")
        # soup("a")
        #print(url2)
        browser.get(url2)
        time.sleep(150)
        html_page2 = browser.page_source
        soup2 = BeautifulSoup(html_page2)
        #print(soup2.prettify())
        from selenium.webdriver.common.by import By
        try:
            button2 = browser.find_element(By.CLASS_NAME,'commit')
            for e in button2:
                e.click()
            #button2.click()
        except NoSuchElementException:
            exceptlist.append(url2)
```
